**General Questions**

```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.
print(data.shape)
```

```python
#Find the average age of students.
avg_age = np.mean(data[:, 0])
avg_age
```

```python
#Extract Math marks of all students.
ext_math = (data[:, 1])
ext_math
```

```python
#Find the highest Science mark.
highest_science = np.max(data[:,2])
(highest_science)
```

```python
#Get details of the student who scored more than 90 in Math.
ext_math1 = ext_math >90
std = data[ext_math1]
std
```

```python
#Increase Math marks of all students by 5.
ext_math +=5
ext_math
```

```python
#Find how many students are younger than 19.
count = np.sum(data[:, 0] < 19)
count
```

```python
#Calculate the average marks in each subject (column-wise mean).
# avg_math = np.mean(data[:, 1])
# avg_sci = np.mean(data[:, 2])
# avg_math
# avg_sci
avg_sub = np.mean(data[:, 1:], axis = 0)
avg_sub
```

```python
#Get data of students who scored at least 80 in both subjects.
mask = (data[:,1] >= 80) & (data[:,2] >= 80)
result = data[mask]
print(result)
```

```python
#Replace all Science marks < 75 with 0.
data[data[:,2] < 75, 2] = 0
print(data)
```
